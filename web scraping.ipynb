```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.edge.service import Service
```

```python
Path=(r"C:\Users\JMN-UK\Desktop\edgedriver_win64\msedgedriver.exe")
service=Service(Path)
driver=webdriver.Edge(service=service)


```

```python
driver.get("https://www.imdb.com/chart/top/")
```

```python
Title=driver.find_elements(By.XPATH, "//div[@class='ipc-title ipc-title--base ipc-title--title ipc-title-link-no-icon ipc-title--on-textPrimary sc-87337ed2-2 dRlLYG cli-title with-margin']")
for i in Title:
    print(i.text)


```

```python
Year=driver.find_elements(By.XPATH, "//div[@class='sc-b4f120f6-6 kprlzj cli-title-metadata']")
for i in Year:
    print(i.text)
```

```python
Rating=driver.find_elements(By.XPATH, "//span[@class='ipc-rating-star--rating']")
for i in Rating:
    print(i.text)
```

```python
Rank=driver.find_elements(By.XPATH, "//div[@class='ipc-signpost__text']")
for i in Rank:
    print(i.text)
```

```python
import csv

with open("Top 250 Movies.csv","w",encoding="utf-8", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Rank", "Title", "Year", "Rating"])
    for i in range(min(len(Title), len(Year),len(Rating),len(Rank))):
        writer.writerow([i+1, Title[i].text, Year[i].text, Rating[i].text])
```
